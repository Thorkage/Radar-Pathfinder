# Converting dataflashlogs from ardupilot to csv 

All data fields can be found at this link:
https://ardupilot.org/copter/docs/logmessages.html#logmessages

The most important ones are:
- BARO (Barometric sensors):
    - Alt

- CTUN (Results of EKF):
    - Alt: 
    - SAlt: Rangefinder (laser) altitude (distance)

- GPS (GNSS variables):
    - Status:  0 - no GPS | 1 - no Fix | 2 - 2D Fix | 3 - 3D Fix (not RTK!) | 4 - DPGS | 5 - RTK Float | 6 - RTK FIX (this is desired)
    - NSats: Number of satellites used
    - HDop: Horizontal dilution of precision
    - Lat
    - Lng
    - Alt: GNSS Altitude

- GPA (GNSS quality variables):
    - VAcc: Vertical accuracy
    - HAcc: Horizontal accuracy
    - VDop: Vertical dilution of precision (https://novotech.com/pages/vertical-dilution-of-precision-vdop)

- POS 
    - Lat: "Canonical" Latitude
    - Lng: "Canonical" Longitude
    - Alt: "Canonical" Altitude

- ATT
    - Roll: The vehicle’s actual roll in degrees (roll left is negative, right is positive)
    - Pitch: The vehicle’s actual pitch angle in degrees (pitch forward is negative, pitch back is positive)
    - Yaw: The vehicle’s actual heading in degrees with 0 = north

```python
import pandas as pd
import numpy as np
from pyproj import Transformer, CRS
import utm
```

```python
crucial_cols = ['GPS.Lat', 'GPS.Lng'] #if one those cols is all nan, we skip the processing of the file


flash_log_path = '/Volumes/PortableSSD/UWIBASS/dataflashlogs/00000067_processed.csv'
df_dataflash = pd.read_csv(flash_log_path)

dropped_cols = [col for col in df_dataflash.columns if df_dataflash[col].isna().all()]


# Check if any crucial columns are being dropped
if any(col in dropped_cols for col in crucial_cols):
    print(f"Skipping file {flash_log_path} because crucial columns are all NaN: {set(dropped_cols) & set(crucial_cols)}")
    raise ValueError(f"Crucial columns {set(dropped_cols) & set(crucial_cols)} are all NaN in {flash_log_path}")

df_dataflash = df_dataflash.dropna(axis=1, how='all')
df_dataflash = df_dataflash.interpolate(method='linear').dropna()

utm_number = utm.from_latlon(df_dataflash['GPS.Lat'].iloc[0], df_dataflash['GPS.Lng'].iloc[0])[2]

transformer = Transformer.from_crs("epsg:4326", CRS.from_proj4(f"+proj=utm +zone={utm_number} +ellps=WGS84 +datum=WGS84 +units=m +type=crs"), always_xy=True)
reverse_transformer = Transformer.from_crs(CRS.from_proj4(f"+proj=utm +zone={utm_number} +ellps=WGS84 +datum=WGS84 +units=m +type=crs"), "epsg:4326", always_xy=True)

df_dataflash['UTM_x'], df_dataflash['UTM_y'] = transformer.transform(df_dataflash['GPS.Lng'].values, df_dataflash['GPS.Lat'].values)

df_dataflash['cumulative_distance'] = np.sqrt(
    np.diff(df_dataflash['UTM_x'], prepend=df_dataflash['UTM_x'].iloc[0])**2 +
    np.diff(df_dataflash['UTM_y'], prepend=df_dataflash['UTM_y'].iloc[0])**2
).cumsum()

df_dataflash['dist_from_start'] = np.sqrt(
    (df_dataflash['UTM_x'] - df_dataflash['UTM_x'].iloc[0])**2 +
    (df_dataflash['UTM_y'] - df_dataflash['UTM_y'].iloc[0])**2
)

df_dataflash['Time'] = df_dataflash['Time'].apply(lambda t: t if '.' in t else t + '.000000')

df_dataflash['timestamp'] = pd.to_datetime(df_dataflash['Date'] + ' ' + df_dataflash['Time'])

df_dataflash.to_csv(flash_log_path, index=False)
```

```python
import re
import os

dir = '/Volumes/PortableSSD/SnowDrone/UWiBaSS/'

# List all folders in the directory that match the pattern XX_XX_XXXX_*
pattern = re.compile(r'^\d{2}_\d{2}_\d{4}_.*$')
datasets = sorted([f for f in os.listdir(dir) if os.path.isdir(os.path.join(dir, f)) and pattern.match(f)])
existing_datasets = []


# datasets = ['28_04_2022_1']

for dataset in datasets:
    if dataset == '01_04_2019_2' or dataset == '01_04_2019_3':
        print(f"Skipping dataset {dataset}. I broke it :(")
        continue
    
    full_path = os.path.join(dir, dataset, 'uwibass_object.pkl')
    if not os.path.exists(full_path):
        print(f"uwibass object does not exist: {full_path}. Skipping {dataset}.")
        continue
    
    with open(full_path, 'rb') as inp:
        uwibass = pickle.load(inp)
    uwibass.dataset_name = dataset
    
    
    #? I am not sure why exactly i am doing the outlier detection 
    # Detect outliers in uwibass.timestamp (difference > 1 hour from previous)
    timestamps = np.array(uwibass.timestamp.flatten())
    # Calculate time differences in seconds
    diffs = np.diff(timestamps)
    # Outlier indices: where the difference to previous is > 3600 seconds (1 hour)
    outlier_idx = np.where(np.abs(diffs) > 3600)[0] + 1  # +1 because diff is between i and i-1

    # Also check for first value being an outlier (compared to next)
    if len(timestamps) > 1 and abs(timestamps[1] - timestamps[0]) > 3600:
        outlier_idx = np.insert(outlier_idx, 0, 0)

    # Also check for last value being an outlier (compared to previous)
    if len(timestamps) > 1 and abs(timestamps[-1] - timestamps[-2]) > 3600:
        outlier_idx = np.append(outlier_idx, len(timestamps) - 1)

    # Replace outliers with linear extrapolation from adjacent values
    for idx in outlier_idx:
        if idx == 0:
            # Extrapolate from next two values
            if len(timestamps) > 2:
                timestamps[idx] = 2 * timestamps[1] - timestamps[2]
            else:
                timestamps[idx] = timestamps[1]  # fallback
        elif idx == len(timestamps) - 1:
            # Extrapolate from previous two values
            if len(timestamps) > 2:
                timestamps[idx] = 2 * timestamps[-2] - timestamps[-3]
            else:
                timestamps[idx] = timestamps[-2]  # fallback
        else:
            # Linear interpolation between neighbors
            timestamps[idx] = (timestamps[idx - 1] + timestamps[idx + 1]) / 2

    uwibass.timestamp = timestamps
    uwibass.datetime_timestamp = [datetime.fromtimestamp(ts) for ts in uwibass.timestamp]

    uwibass, path_dataflashlog = find_dataflashlog(uwibass)
    
    if not path_dataflashlog:
        continue
    
    uwibass.dataflashlog_path = path_dataflashlog

    df_dataflashlog = load_dataflashlog(uwibass)

    uwibass = attach_dataflashlog(uwibass, df_dataflashlog,
                                    columns=['GPS.Lat', 'GPS.Lng', 'GPS.Status', 'GPS.Alt',
                                            'CTUN.Alt', 'CTUN.SAlt', 'POS.Alt', 'BARO.Alt',
                                            'UTM_x', 'UTM_y', 
                                            ]
                                    )
    
    with open(full_path, 'wb') as outp:
            pickle.dump(uwibass, outp, pickle.HIGHEST_PROTOCOL)
    existing_datasets.append(dataset)
    
    print(f"Dataflashlog added: {uwibass.dataflashlog_path} to {full_path}")
```
